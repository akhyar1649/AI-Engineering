# Solution — RAG Optimization

```python
def metrics(relevant, retrieved):
    relevant=set(relevant); retrieved=list(retrieved)
    hits=len(relevant & set(retrieved))
    return {'precision':hits/len(retrieved),'recall':hits/len(relevant),'mrr':next((1/(i+1) for i,x in enumerate(retrieved) if x in relevant),0)}
metrics(['a','c'],['b','a','d'])
```

Metric function menjadi dasar regression test; data production harus dianonimkan dan memiliki label yang dapat direview.
