```python
class QueuePolicy:
    def __init__(self, max_inflight=2): self.max_inflight=max_inflight; self.inflight=0
    def accept(self):
        if self.inflight>=self.max_inflight: return {'accepted':False,'reason':'queue_limit'}
        self.inflight+=1; return {'accepted':True}
    def finish(self): self.inflight=max(0,self.inflight-1)
q=QueuePolicy(); [q.accept() for _ in range(3)]
```
