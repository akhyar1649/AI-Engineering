# Prompt Technique and Optimization — PT Kirimin

```python
def build_prompt(question, evidence):
    return {'system':'Anda adalah assistant SOP Kirimin. Jawab hanya dari evidence dan eskalasi jika kurang.', 'user':question, 'evidence':evidence, 'output_schema':['answer','citations','needs_human']}
build_prompt('Bagaimana mengubah alamat?', [{'id':'SOP-12','text':'Perubahan alamat hanya sebelum pickup.'}])
```

```python
def mock_score(prompt):
    required=['evidence','output_schema']
    return sum(key in prompt for key in required)/len(required)
mock_score(build_prompt('x',[]))
```
