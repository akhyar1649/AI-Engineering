# RAG and LangChain — PT Kirimin
Kita memisahkan retrieval, prompt construction, dan generation mock.

> Diverifikasi: LangChain Python v1.x — sumber: https://docs.langchain.com/oss/python/releases/langchain-v1 — tanggal cek: 2026-09-18

```python
docs=[{'id':'SOP-12','text':'Perubahan alamat hanya sebelum pickup.'},{'id':'SOP-13','text':'Refund memerlukan review finance.'}]
def retrieve(query, k=2):
    q=set(query.lower().split())
    return sorted(docs,key=lambda d:len(q & set(d['text'].lower().split())),reverse=True)[:k]
def build_context(found): return '\n'.join(f"[{d['id']}] {d['text']}" for d in found)
def mock_generate(question, context):
    return {'answer':f'Jawab berdasarkan context: {context.splitlines()[0]}','citations':[context.split(']')[0]+']']}
found=retrieve('ubah alamat pickup')
mock_generate('ubah alamat',build_context(found))
```
