# Solution — Fine-Tuning Decision

```python
def choose_strategy(knowledge_changes, style_gap, retrieval_gap):
    if retrieval_gap: return 'fix_retrieval'
    if knowledge_changes: return 'RAG'
    if style_gap: return 'fine_tune_candidate'
    return 'prompt_baseline'
[choose_strategy(True,False,False),choose_strategy(False,True,False),choose_strategy(False,False,False)]
```

Decision gate menghindari fine-tuning ketika masalah sebenarnya ada pada knowledge freshness atau retrieval.
