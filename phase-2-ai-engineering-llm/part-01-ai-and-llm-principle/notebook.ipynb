Kita membuat baseline model contract dengan mock provider agar tidak memerlukan API key.

> Diverifikasi: OpenAI Python SDK v3.15.0 — sumber: https://github.com/openai/openai-python/releases — tanggal cek: 2026-09-18
> Diverifikasi: Gemini stable model — sumber: https://ai.google.dev/gemini-api/docs/models — tanggal cek: 2026-09-18

```python
def approx_tokens(text):
    return max(1, len(text.split()))
def mock_completion(question, context='', temperature=0.0):
    return {'answer': 'Gunakan SOP pengiriman dan eskalasi ke agent bila evidence tidak cukup.', 'input_tokens':approx_tokens(question+' '+context), 'temperature':temperature}
mock_completion('Bagaimana mengubah alamat kiriman?', 'SOP address change v2')
```
