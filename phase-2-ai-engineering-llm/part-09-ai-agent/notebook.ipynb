```python
tools={'tracking':lambda shipment_id:{'status':'in_transit','shipment_id':shipment_id},'sop':lambda topic:{'source':'SOP-12','text':'Perubahan alamat sebelum pickup.'}}
def agent(intent, args, max_steps=3):
    if max_steps<=0: return {'status':'stopped','reason':'step_limit'}
    if intent not in tools: return {'status':'human_review','reason':'no_safe_tool'}
    return {'status':'done','tool':intent,'result':tools[intent](**args)}
agent('tracking',{'shipment_id':'S-1'}), agent('refund',{})
```
