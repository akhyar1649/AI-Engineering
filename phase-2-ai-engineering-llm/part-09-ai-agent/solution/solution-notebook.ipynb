# Solution — AI Agent

```python
def route(intent):
    policy={'tracking':'tracking','faq':'sop','refund':'human_review','unknown':'human_review'}
    return policy.get(intent,'human_review')
[route(x) for x in ['tracking','faq','refund','unknown']]
```

Routing policy menjadi guardrail deterministik sebelum agent memilih action.
