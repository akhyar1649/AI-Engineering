# Solution — Structured RAG

```python
templates={'urgent_open_tickets':'SELECT COUNT(*) AS count FROM customer_tickets WHERE priority = ? AND resolved_at IS NULL'}
def build_query(intent, params):
    return {'sql':templates[intent],'params':params,'provenance':{'intent':intent,'timezone':'Asia/Jakarta'}}
build_query('urgent_open_tickets',['urgent'])
```

Template menjaga query shape dan memisahkan value parameter dari SQL string.
