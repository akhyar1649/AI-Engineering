```python
allowed={'customer_tickets','orders'}
def validate_sql(sql):
    normalized=sql.lower().strip()
    forbidden=['delete','update','insert','drop','alter',';']
    return normalized.startswith('select') and not any(word in normalized for word in forbidden) and any(table in normalized for table in allowed)
[validate_sql(x) for x in ['SELECT COUNT(*) FROM customer_tickets','DELETE FROM orders','SELECT * FROM secret_table']]
```
