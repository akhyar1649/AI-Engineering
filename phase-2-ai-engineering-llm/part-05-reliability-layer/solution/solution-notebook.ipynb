# Solution — Reliability Layer

```python
def call_with_policy(outcomes, max_attempts=2):
    for attempt, outcome in enumerate(outcomes[:max_attempts], start=1):
        if outcome=='ok': return {'status':'success','attempts':attempt}
        if outcome not in {'timeout','rate_limit'}: return {'status':'failed','reason':outcome}
    return {'status':'fallback','reason':'retry_budget_exhausted'}
call_with_policy(['timeout','ok']), call_with_policy(['invalid'])
```

Wrapper mengembalikan status terstruktur sehingga caller tidak menebak dari string error.
