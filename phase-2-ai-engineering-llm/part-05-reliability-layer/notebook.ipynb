```python
def validate(response):
    required={'answer','citations','needs_human'}
    return required <= response.keys() and isinstance(response['citations'],list)
def policy(call, attempts=0):
    if call=='timeout' and attempts<2: return {'action':'retry','attempt':attempts+1}
    if call in {'timeout','rate_limit'}: return {'action':'fallback','reason':'provider_unavailable'}
    if call=='sensitive': return {'action':'human_review','reason':'policy_boundary'}
    return {'action':'accept'}
[policy(x) for x in ['timeout','rate_limit','sensitive','ok']]
```
