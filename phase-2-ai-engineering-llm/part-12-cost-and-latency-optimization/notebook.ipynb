# Cost & Latency Optimization — PT Kirimin

```python
prices={'fast_input':0.0001,'fast_output':0.0003,'strong_input':0.001,'strong_output':0.003}
def cost(route,input_tokens,output_tokens):
    prefix='fast' if route=='fast' else 'strong'
    return input_tokens*prices[f'{prefix}_input']+output_tokens*prices[f'{prefix}_output']
cost('fast',500,100), cost('strong',500,100)
```

```python
requests=['faq','tracking','refund','faq','faq']
cache=set(); hits=0
for request in requests:
    if request in cache: hits+=1
    cache.add(request)
hits/len(requests)
```
