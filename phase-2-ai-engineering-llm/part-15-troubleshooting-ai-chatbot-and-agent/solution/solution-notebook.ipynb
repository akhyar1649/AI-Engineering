# Solution — AI Troubleshooting

```python
def mitigation(stage):
    actions={'policy':'disable_route','retrieval':'reindex_or_fix_filter','provider':'fallback_provider','output':'reject_and_review','tool':'remove_permission'}
    return actions.get(stage,'collect_sanitized_trace')
[mitigation(x) for x in ['policy','retrieval','provider','output','tool']]
```

Mitigasi harus spesifik terhadap stage agar incident tidak hanya ditutup dengan retry.
