# Solution — AI Data Infrastructure

```python
chunks=[{'id':'a','text':'refund finance review','access':'support'},{'id':'b','text':'refund internal finance','access':'finance'}]
def retrieve(query, role):
    candidates=[c for c in chunks if c['access'] in {role,'public'}]
    terms=set(query.split())
    return sorted(candidates,key=lambda c:len(terms & set(c['text'].split())),reverse=True)
retrieve('refund finance','support')
```

Access filter diterapkan sebelum ranking dan generation.
