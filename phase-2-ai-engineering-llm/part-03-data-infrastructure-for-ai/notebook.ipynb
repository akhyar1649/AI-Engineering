```python
documents=[{'id':'SOP-1-v2','text':'Perubahan alamat hanya sebelum pickup.','version':2,'effective':'2026-09-01','access':'support'}, {'id':'SOP-2-v1','text':'Refund memerlukan review finance.','version':1,'effective':'2026-08-01','access':'support'}]
chunks=[{'chunk_id':f"{doc['id']}-0",'text':doc['text'],'metadata':{k:doc[k] for k in ['id','version','effective','access']}} for doc in documents]
chunks
```

```python
def token_set(text): return set(text.lower().replace('.','').split())
def jaccard(query, text):
    a,b=token_set(query),token_set(text)
    return len(a&b)/len(a|b) if a|b else 0
[(c['chunk_id'],jaccard('ubah alamat sebelum pickup',c['text'])) for c in chunks]
```
