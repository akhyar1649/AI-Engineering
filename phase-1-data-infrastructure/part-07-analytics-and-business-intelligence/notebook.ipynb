Kita memulai dari definisi metric, lalu membuat data yang siap divisualisasikan.

> Diverifikasi: Tableau release notes — sumber: https://help.tableau.com/current/tableau/en-us/whatsnew_all.htm — tanggal cek: 2026-09-18

```python
import pandas as pd
orders=pd.DataFrame({'city':['Jakarta','Jakarta','Bandung','Bandung'],'paid':[1,1,1,0],'delivered':[1,0,1,0]})
eligible=orders.query('paid == 1')
kpi=(eligible.groupby('city',as_index=False).agg(eligible_orders=('paid','size'),delivered_orders=('delivered','sum')))
kpi['delivery_success_rate']=kpi['delivered_orders']/kpi['eligible_orders']
kpi
```

Tambahkan kolom `freshness_at`, `definition_version`, dan `data_quality_status` pada serving dataset agar dashboard memiliki konteks.
