# Solution — Advanced SQL

```python
import sqlite3, pandas as pd
cx=sqlite3.connect(':memory:')
events=pd.DataFrame([('O-1','created','2026-09-01 08:00'),('O-1','picked_up','2026-09-01 10:00'),('O-1','delivered','2026-09-02 12:00'),('O-2','created','2026-09-01 09:00')],columns=['order_id','status','event_time'])
events.to_sql('events',cx,index=False)
pd.read_sql_query('''WITH x AS (SELECT *, LAG(event_time) OVER(PARTITION BY order_id ORDER BY event_time) AS previous_time FROM events) SELECT *, CASE WHEN previous_time IS NOT NULL AND event_time < previous_time THEN 1 ELSE 0 END AS out_of_order FROM x''',cx)
```

Solusi memisahkan deteksi anomali dari business metric sehingga metrik dapat dibaca bersama quality signal.
