```python
def quality_gate(rows, invalid_threshold=0.05):
    invalid=[row for row in rows if not row.get('order_id') or row.get('order_value_idr',0)<0]
    rate=len(invalid)/len(rows) if rows else 1
    return {'input_rows':len(rows),'invalid_rows':len(invalid),'invalid_rate':rate,'decision':'quarantine' if rate>invalid_threshold else 'pass'}
quality_gate([{'order_id':'O-1','order_value_idr':100},{'order_id':None,'order_value_idr':10}])
```

Gunakan output ini sebagai starting point untuk test quality gate, bukan pengganti integration test pada environment deployment.
