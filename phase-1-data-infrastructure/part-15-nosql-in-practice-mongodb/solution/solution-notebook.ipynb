# Solution — MongoDB Practice

```python
def ingest(events):
    store={}
    for e in events:
        if e['event_id'] in store: continue
        if e['status'] not in {'created','in_transit','delivered','failed'}: continue
        store[e['event_id']]=e
    latest={}
    for e in store.values():
        if e['shipment_id'] not in latest or e['event_time']>latest[e['shipment_id']]['event_time']:
            latest[e['shipment_id']]=e
    return list(latest.values())
ingest([{'event_id':'E-1','shipment_id':'S-1','status':'created','event_time':'2026-09-18T08:00'},{'event_id':'E-1','shipment_id':'S-1','status':'created','event_time':'2026-09-18T08:00'}])
```

Operasi dedupe dan latest dipisahkan agar audit event tetap tersedia.
