```python
events=[
 {'event_id':'E-1','shipment_id':'S-1','status':'created','event_time':'2026-09-18T08:00:00'},
 {'event_id':'E-2','shipment_id':'S-1','status':'in_transit','event_time':'2026-09-18T09:00:00'},
 {'event_id':'E-2','shipment_id':'S-1','status':'in_transit','event_time':'2026-09-18T09:00:00'},
]
unique={event['event_id']:event for event in events}
latest={}
for event in unique.values():
    current=latest.get(event['shipment_id'])
    if current is None or event['event_time']>current['event_time']: latest[event['shipment_id']]=event
list(latest.values())
```
