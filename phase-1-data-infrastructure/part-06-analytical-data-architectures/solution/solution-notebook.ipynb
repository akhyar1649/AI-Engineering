# Solution — Analytical Data Architectures

```python
import pandas as pd
fact=pd.DataFrame({'order_id':['O-1','O-2','O-3'],'order_date':pd.to_datetime(['2026-09-01','2026-09-01','2026-09-02']),'city':['Jakarta','Bandung','Jakarta'],'value_idr':[100,200,300]})
fact['date_key']=fact['order_date'].dt.strftime('%Y%m%d').astype(int)
fact.groupby(['date_key','city'],as_index=False)['value_idr'].sum()
```

Output dapat dipartisi berdasarkan `date_key`, sementara city menjadi dimension untuk filtering.
