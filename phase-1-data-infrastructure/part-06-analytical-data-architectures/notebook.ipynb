Kita mengubah event order menjadi fact table dengan dimensi kota dan tanggal.

> Diverifikasi: PostgreSQL v18.6 — sumber: https://www.postgresql.org/docs/release/ — tanggal cek: 2026-09-18

```python
import pandas as pd
orders=pd.DataFrame({'order_id':['O-1','O-2','O-3'],'city':['Jakarta','Bandung','Jakarta'],'order_date':pd.to_datetime(['2026-09-01','2026-09-01','2026-09-02']),'value_idr':[100,200,300]})
dim_city=orders[['city']].drop_duplicates().reset_index(names='city_key')
fact=orders.merge(dim_city,on='city').rename(columns={'order_id':'order_key'})
dim_city, fact
```

```python
daily=(fact.groupby(['order_date','city_key'],as_index=False)['value_idr'].sum())
daily
```
