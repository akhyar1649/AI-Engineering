# Solution — Data Cleaning and Processing

```python
import pandas as pd
def clean_orders(raw):
    out=raw.copy()
    out['city_clean']=out['origin_city'].astype('string').str.strip().str.lower().replace({'jkt':'jakarta'}).str.title()
    out['created_clean']=pd.to_datetime(out['order_created_at'], dayfirst=True, errors='coerce')
    out['value_clean']=pd.to_numeric(out['order_value_idr'], errors='coerce')
    mask=out['order_id'].isna() | out['created_clean'].isna() | out['value_clean'].isna() | (out['value_clean']<0) | (out['value_clean']>10_000_000)
    quarantine=out.loc[mask].copy()
    quarantine['reason']='domain or parse validation failed'
    curated=out.loc[~mask].sort_values('created_clean').drop_duplicates('order_id',keep='last').reset_index(drop=True)
    return curated, quarantine

sample=pd.DataFrame({'order_id':['O-1','O-1',None],'origin_city':['JKT','Jakarta','Bandung'],'order_created_at':['01/09/2026','2026-09-02','bad'],'order_value_idr':['100','120','x']})
clean_orders(sample)
```
