```python
events_per_day=80_000
bytes_per_event=1_200
retention_days=90
replication=3
headroom=1.3
payload_gb=events_per_day*bytes_per_event*retention_days*replication/1_000_000_000
round(payload_gb*headroom,2)
```

```python
scenarios={'baseline':1,'growth_2x':2,'growth_5x':5}
{key:round(payload_gb*factor*headroom,2) for key,factor in scenarios.items()}
```
