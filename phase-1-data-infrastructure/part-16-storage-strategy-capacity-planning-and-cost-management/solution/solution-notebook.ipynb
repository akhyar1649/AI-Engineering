# Solution — Storage Capacity

```python
def capacity(events_per_day, bytes_per_event, days, replication=3, overhead=1.3, compression=0.5):
    return events_per_day*bytes_per_event*days*replication*overhead*compression/1_000_000_000
{factor:round(capacity(80_000*factor,1_200,90),2) for factor in [1,2,5]}
```

Fungsi menempatkan asumsi sebagai parameter sehingga scenario mudah diuji.
