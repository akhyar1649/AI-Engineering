# Solution — Reliability Prioritization

```python
incidents=[('source_timeout',12000,40),('quality_failure',12000,10),('stale_dashboard',5000,90),('single_bad_row',1,0)]
for name,orders,late in incidents:
    severity='P1' if orders>=10000 or late>=60 else ('P2' if orders>0 else 'P3')
    print(name,severity)
```

Klasifikasi menggabungkan customer impact dan lateness; angka threshold harus disepakati dalam SLO.
