# Solution — Data Governance Fundamentals & Quality Management

```python
import pandas as pd
df=pd.DataFrame({'order_id':['O-1','O-2','O-2',None],'order_value_idr':[100,200,-5,300]})
rules=[('not_null_order_id',df['order_id'].notna().all(),'blocking'),('unique_order_id',df['order_id'].dropna().is_unique,'blocking'),('non_negative_value',(df['order_value_idr']>=0).all(),'blocking')]
result=[{'name':n,'passed':bool(p),'severity':s} for n,p,s in rules]
result
```

Validation result menyimpan nama rule, status, dan severity sehingga orchestrator dapat menerapkan policy.
