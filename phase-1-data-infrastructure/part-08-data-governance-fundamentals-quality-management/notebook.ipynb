# Data Governance Fundamentals & Quality Management
Kita membuat quality checks sebagai kode dan mengubah hasilnya menjadi keputusan pipeline.

> Diverifikasi: Great Expectations v1.23.0 — sumber: https://docs.greatexpectations.io/docs/core/changelog/ — tanggal cek: 2026-09-18

```python
import pandas as pd
df=pd.DataFrame({'order_id':['O-1','O-2','O-2',None],'order_value_idr':[100,200,-5,300]})
checks={
 'order_id_not_null': bool(df['order_id'].notna().all()),
 'order_id_unique': bool(df['order_id'].dropna().is_unique),
 'value_non_negative': bool((df['order_value_idr']>=0).all()),
}
checks
```

```python
blocking={'order_id_not_null','value_non_negative'}
failed={name for name, passed in checks.items() if not passed}
decision='quarantine' if failed & blocking else 'pass'
{'failed':failed,'decision':decision}
```
