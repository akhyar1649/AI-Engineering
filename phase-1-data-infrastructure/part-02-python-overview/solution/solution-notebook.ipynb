# Solution — Python Overview

```python
import pandas as pd
def transform_orders(frame):
    required = {'order_id','origin_city','order_value_idr'}
    missing = required - set(frame.columns)
    if missing: raise ValueError(f'missing columns: {sorted(missing)}')
    out = frame.copy()
    out['origin_city'] = out['origin_city'].astype('string').str.strip().str.lower().replace({'jkt':'jakarta'}).str.title()
    out['order_value_idr'] = pd.to_numeric(out['order_value_idr'], errors='coerce')
    errors = out['order_id'].isna() | out['order_value_idr'].isna() | (out['order_value_idr'] < 0)
    report = {'input_rows': len(out), 'rejected_rows': int(errors.sum()), 'duplicates': int(out['order_id'].duplicated(keep=False).sum())}
    out = out.loc[~errors].drop_duplicates('order_id', keep='last').reset_index(drop=True)
    return out, report

sample = pd.DataFrame({'order_id':['O-1','O-1','O-2'], 'origin_city':['JKT','Jakarta','Bandung'], 'order_value_idr':['100','110','bad']})
transform_orders(sample)
```
