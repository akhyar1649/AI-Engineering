# Solution — Database and Big Data Pipeline

```python
from dataclasses import dataclass
@dataclass
class PartitionStore:
    curated: dict
    quarantine: dict
    def write(self, run_date, rows):
        valid=[r for r in rows if r.get('order_id') and r.get('order_value_idr',0)>=0]
        invalid=[r for r in rows if r not in valid]
        self.curated[run_date]=valid
        self.quarantine[run_date]=invalid
store=PartitionStore({}, {})
store.write('2026-09-18',[{'order_id':'O-1','order_value_idr':10},{'order_id':None,'order_value_idr':2}])
store.write('2026-09-18',[{'order_id':'O-1','order_value_idr':10}])
store.curated
```

Menulis berdasarkan partition key membuat retry menggantikan hasil partition yang sama, bukan append buta.
