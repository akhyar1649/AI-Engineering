# Solution — Containerization & Deployment

```python
import os
def readiness():
    required=['QUALITY_STORE_URL']
    missing=[key for key in required if not os.getenv(key)]
    return {'status':'ready' if not missing else 'not_ready','missing':missing}
readiness()
```

Readiness memeriksa dependency yang benar-benar diperlukan; liveness cukup memeriksa process.
