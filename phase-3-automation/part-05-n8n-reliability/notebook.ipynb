```python
def classify(error):
    if error in {'timeout','rate_limit','5xx'}: return 'retry'
    if error in {'validation','auth','business_rejection'}: return 'dead_letter'
    return 'alert'
[classify(x) for x in ['timeout','validation','auth','unknown']]
```

```python
processed=set()
def dedup(event_id):
    if event_id in processed: return False
    processed.add(event_id); return True
[dedup(x) for x in ['E-1','E-1','E-2']]
```
