# Solution — n8n Reliability

```python
def execution(event, store, attempts=0):
    if event['event_id'] in store['processed']: return {'status':'duplicate'}
    if event.get('error')=='timeout' and attempts<2: return {'status':'retry','attempt':attempts+1}
    if event.get('error'): store['dlq'].append(event); return {'status':'dead_letter'}
    store['processed'].add(event['event_id']); return {'status':'success'}
store={'processed':set(),'dlq':[]}
execution({'event_id':'E-1'},store), execution({'event_id':'E-1'},store)
```

Dedupe dilakukan sebelum side effect dan DLQ mempertahankan input untuk investigasi/replay.
