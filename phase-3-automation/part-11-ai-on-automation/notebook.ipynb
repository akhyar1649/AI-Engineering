```python
def classify(message):
    text=message.lower()
    if 'refund' in text: return {'label':'refund','confidence':0.92}
    if 'status' in text: return {'label':'tracking','confidence':0.88}
    return {'label':'unknown','confidence':0.45}
def route(message):
    result=classify(message)
    return {'result':result,'next':'human_review' if result['label']=='refund' or result['confidence']<0.7 else 'automated_flow'}
[route(x) for x in ['cek status order','saya minta refund','tolong bantu']]
```
