# Solution — MCP Capability Policy

```python
def invoke(capability, scope, approved=False):
    if scope!=capability['scope']: return {'ok':False,'reason':'scope_denied'}
    if capability['side_effect'] and not approved: return {'ok':False,'reason':'approval_required'}
    return {'ok':True,'audit':capability['name']}
invoke({'name':'refund','scope':'finance:write','side_effect':True},'finance:write'), invoke({'name':'refund','scope':'finance:write','side_effect':True},'finance:write',True)
```

Scope dan approval diperiksa sebelum tool dipanggil.
