# Capstone 3 Reference Checks

```python
def process(event, state):
    key=event['event_id']
    if key in state['seen']: return {'status':'duplicate_prevented'}
    state['seen'].add(key)
    if event.get('status')=='failed':
        state['dlq'].append(event); return {'status':'dead_letter'}
    state['ledger'].append(event); return {'status':'persisted'}
state={'seen':set(),'ledger':[],'dlq':[]}
process({'event_id':'E-1','status':'delivered'},state), process({'event_id':'E-1','status':'delivered'},state)
```

Reference checks menekankan durable ledger, duplicate prevention, dan DLQ.
