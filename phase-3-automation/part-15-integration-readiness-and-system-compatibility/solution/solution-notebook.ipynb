# Solution — Integration Readiness

```python
def go_no_go(checks, blockers):
    missing=[name for name,value in checks.items() if not value]
    blocking=[name for name in missing if name in blockers]
    return {'decision':'go' if not blocking else 'no_go','missing':missing,'blocking':blocking}
go_no_go({'auth':True,'idempotency':False,'owner':True},{'idempotency'})
```

Blocker list membuat keputusan aman meski ada non-blocking gap yang dapat di-remediasi.
