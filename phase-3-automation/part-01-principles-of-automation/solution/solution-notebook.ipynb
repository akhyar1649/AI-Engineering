# Solution — Automation Fit

```python
def decision(volume, deterministic, risk):
    if risk>=4: return 'human_review'
    return 'automate' if volume>1000 and deterministic else 'pilot'
[decision(10000,True,2),decision(100,True,5),decision(500,False,2)]
```

Decision memasukkan risk boundary sebelum ROI.
