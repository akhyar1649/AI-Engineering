# Solution — n8n Data Logic

```python
def process(events, max_items=100):
    seen=set(); output=[]
    for event in events[:max_items]:
        key=event.get('event_id')
        if not key or key in seen: continue
        seen.add(key)
        output.append({'correlation_id':key,'route':'notify' if event.get('status')=='delivered' else 'exception'})
    return output
process([{'event_id':'E-1','status':'delivered'},{'event_id':'E-1','status':'delivered'}])
```

Limit dan dedup membuat loop bounded dan retry-safe.
