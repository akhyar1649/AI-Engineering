# Solution — Automation Observability

```python
events=[{'correlation_id':'C-1','stage':'ledger','status':'success'},{'correlation_id':'C-1','stage':'notification','status':'accepted'},{'correlation_id':'C-2','stage':'ledger','status':'success'},{'correlation_id':'C-2','stage':'notification','status':'failed'}]
from collections import defaultdict
states=defaultdict(list)
for e in events: states[e['correlation_id']].append(e)
{key:[x['status'] for x in value] for key,value in states.items()}
```

Business outcome harus dilihat lintas stage, bukan dari satu execution status.
