# Solution — Integrated System

```python
def integrate(event, notification_ok=True):
    ledger={'event_id':event['event_id'],'status':event['status']}
    if event['status']=='delivered' and not notification_ok:
        return {'ledger':ledger,'compensation':{'type':'notify_retry','event_id':event['event_id']}}
    return {'ledger':ledger,'compensation':None}
integrate({'event_id':'E-1','status':'delivered'},False)
```

Compensation memperbaiki langkah yang gagal tanpa mengubah fakta event yang sudah tersimpan.
