# Solution — Self-Hosted Readiness

```python
def readiness(checks):
    blocking=[k for k,v in checks.items() if not v]
    return {'ready':not blocking,'blocking':blocking}
readiness({'tls':True,'backup':True,'restore_test':False,'secret_management':True})
```

Restore test adalah blocking check karena backup tanpa restore evidence belum membuktikan recovery.
