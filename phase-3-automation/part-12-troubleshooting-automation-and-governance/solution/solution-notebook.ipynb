# Solution — Automation Incident Response

```python
def containment(event):
    if event['type']=='unauthorized': return ['disable_capability','preserve_audit','notify_owner']
    if event['type']=='duplicate': return ['pause_workflow','dedup_check','replay_dry_run']
    return ['retry_bounded','monitor_dlq']
containment({'type':'unauthorized'}), containment({'type':'duplicate'})
```

Containment harus menghentikan risiko baru sambil mempertahankan bukti untuk diagnosis.
