```python
def validate(event):
    required={'shipment_id','status','event_time'}
    missing=required-set(event)
    return {'valid':not missing,'missing':sorted(missing),'correlation_id':event.get('event_id','generated')}
validate({'shipment_id':'S-1','status':'delivered','event_time':'2026-09-18'})
```

Validation node sebaiknya menghasilkan error yang bisa ditangani, bukan membuat downstream menebak.
