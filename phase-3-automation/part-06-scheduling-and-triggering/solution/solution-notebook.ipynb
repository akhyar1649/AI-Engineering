# Solution — Scheduling & Triggering

```python
def side_effect_allowed(trigger, dry_run=False):
    return trigger=='event' and not dry_run or trigger=='schedule' and not dry_run
[(x,side_effect_allowed(x,False),side_effect_allowed(x,True)) for x in ['event','schedule']]
```

Dry-run mematikan side effect saat backfill; output laporan tetap dapat dibuat.
